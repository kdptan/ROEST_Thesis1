```python
import re

def load_roast_csv(filepath):
    df = pd.read_csv(filepath)

    # Remove leading/trailing spaces from column names
    df.columns = [str(c).strip() for c in df.columns]

    # Standardize Total cracks column name
    df.columns = [
        re.sub(r"^Total cracks \(\d+\)$", "Total cracks", c)
        for c in df.columns
    ]

    return df


# Verify all 120 files now have the same columns
schemas = []

for f in csv_files:
    df = load_roast_csv(f)
    schemas.append(tuple(df.columns))

print("Unique schemas after normalization:", len(set(schemas)))
print("\nStandardized columns:")
print(list(schemas[0]))
```

```python
def get_event_value(df, event_name, column):
    rows = df[df["Event"] == event_name]
    if len(rows) == 0:
        return np.nan
    return rows.iloc[0][column]


raw_reference = []

for f in csv_files:
    df = load_roast_csv(f).copy()
    df["Time_sec"] = df["Time"] / 1000

    batch_id = int(re.search(r"Batch (\d+)", f.name).group(1))

    fc_time = get_event_value(df, "First Crack", "Time_sec")
    drop_time = get_event_value(df, "Drop", "Time_sec")

    raw_reference.append({
        "batch_id": batch_id,
        "file": f.name,

        "charge_temp": get_event_value(
            df, "Charge", "Bean temp (°C)"
        ),

        "tp_time": get_event_value(
            df, "Turning Point", "Time_sec"
        ),
        "tp_temp": get_event_value(
            df, "Turning Point", "Bean temp (°C)"
        ),

        "yellow_time": get_event_value(
            df, "Dry End", "Time_sec"
        ),
        "yellow_temp": get_event_value(
            df, "Dry End", "Bean temp (°C)"
        ),

        "fc_time": fc_time,
        "fc_temp": get_event_value(
            df, "First Crack", "Bean temp (°C)"
        ),

        "drop_time": drop_time,
        "drop_temp": get_event_value(
            df, "Drop", "Bean temp (°C)"
        ),

        "dev_time": drop_time - fc_time
    })

raw_reference = pd.DataFrame(raw_reference)

display(raw_reference.head())

print("Reference roasts:", len(raw_reference))
```

```python
from scipy.optimize import linear_sum_assignment

# Convert DOE time strings such as "8:26.0" into seconds
def time_to_seconds(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    try:
        parts = value.split(":")

        if len(parts) == 2:
            return float(parts[0]) * 60 + float(parts[1])

        return float(value)

    except:
        return np.nan


doe_match = doe.copy()

doe_match["tp_time"] = doe_match["Actual TP Time"].apply(time_to_seconds)
doe_match["yellow_time"] = doe_match["Actual Yellow Time"].apply(time_to_seconds)
doe_match["fc_time"] = doe_match["Actual FC Start Time"].apply(time_to_seconds)
doe_match["drop_time"] = doe_match["Actual Drop Time"].apply(time_to_seconds)


# Features shared by RAW files and DOE
features = [
    ("charge_temp", "Actual Charge Temp (°C)", 1.0),
    ("tp_time", "tp_time", 5.0),
    ("tp_temp", "Actual TP Temp (°C)", 2.0),
    ("yellow_time", "yellow_time", 5.0),
    ("yellow_temp", "Actual Yellow Temp (°C)", 1.0),
    ("fc_time", "fc_time", 5.0),
    ("fc_temp", "Actual FC Temp (°C)", 1.0),
    ("drop_time", "drop_time", 5.0),
    ("drop_temp", "Actual Drop Temp (°C)", 1.0),
]


# Build distance matrix
cost_matrix = np.zeros((len(raw_reference), len(doe_match)))

for i, raw_row in raw_reference.iterrows():
    for j, doe_row in doe_match.iterrows():

        differences = []

        for raw_col, doe_col, scale in features:

            r = raw_row[raw_col]
            d = doe_row[doe_col]

            if pd.notna(r) and pd.notna(d):
                differences.append(((r - d) / scale) ** 2)

        if differences:
            cost_matrix[i, j] = np.mean(differences)
        else:
            cost_matrix[i, j] = 999999


# Force one-to-one matching
raw_indices, doe_indices = linear_sum_assignment(cost_matrix)

matches = []

for ri, di in zip(raw_indices, doe_indices):

    raw_id = int(raw_reference.iloc[ri]["batch_id"])

    old_doe_id = doe_match.iloc[di]["ROEST Roast ID"]
    old_doe_num = int(re.search(r"\d+", str(old_doe_id)).group())

    matches.append({
        "raw_batch_id": raw_id,
        "doe_current_id": old_doe_num,
        "run_number": doe_match.iloc[di]["Run #"],
        "match_score": cost_matrix[ri, di],
        "same_id": raw_id == old_doe_num
    })


matches = pd.DataFrame(matches).sort_values("raw_batch_id")

print("Total matched:", len(matches))
print("Already correct IDs:", matches["same_id"].sum())
print("IDs requiring correction:", (~matches["same_id"]).sum())

print("\nDOE IDs that should be corrected:")
display(
    matches[matches["same_id"] == False][
        ["run_number", "doe_current_id", "raw_batch_id", "match_score"]
    ]
)
```

```python
# Prepare DOE comparison values
doe_verify = doe.copy()

doe_verify["batch_id"] = (
    doe_verify["ROEST Roast ID"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
    .astype("Int64")
)

doe_verify["fc_time"] = doe_verify["Actual FC Start Time"].apply(time_to_seconds)
doe_verify["drop_time"] = doe_verify["Actual Drop Time"].apply(time_to_seconds)
doe_verify["yellow_time"] = doe_verify["Actual Yellow Time"].apply(time_to_seconds)

# Join DOE to RAW using the ID currently written in DOE
check = doe_verify.merge(
    raw_reference,
    on="batch_id",
    how="inner",
    suffixes=("_doe", "_raw")
)

# Calculate differences
check["FC_time_diff"] = abs(check["fc_time_doe"] - check["fc_time_raw"])
check["Drop_time_diff"] = abs(check["drop_time_doe"] - check["drop_time_raw"])
check["Yellow_time_diff"] = abs(check["yellow_time_doe"] - check["yellow_time_raw"])

check["FC_temp_diff"] = abs(
    check["Actual FC Temp (°C)"] - check["fc_temp"]
)

check["Drop_temp_diff"] = abs(
    check["Actual Drop Temp (°C)"] - check["drop_temp"]
)

# Strong agreement test
check["likely_correct"] = (
    (check["FC_time_diff"] <= 2) &
    (check["Drop_time_diff"] <= 2) &
    (check["FC_temp_diff"] <= 0.5) &
    (check["Drop_temp_diff"] <= 0.5)
)

print("IDs existing in BOTH RAW and DOE:", len(check))
print("Strongly matching:", check["likely_correct"].sum())
print("Suspicious:", (~check["likely_correct"]).sum())

print("\nSuspicious existing IDs:")
display(
    check[~check["likely_correct"]][[
        "Run #",
        "batch_id",
        "FC_time_diff",
        "Drop_time_diff",
        "FC_temp_diff",
        "Drop_temp_diff"
    ]]
)
```

```python
raw_missing_ids = sorted(raw_ids - doe_ids)
doe_wrong_ids = sorted(doe_ids - raw_ids)

raw_missing = raw_reference[
    raw_reference["batch_id"].isin(raw_missing_ids)
].reset_index(drop=True)

doe_wrong = doe_verify[
    doe_verify["batch_id"].isin(doe_wrong_ids)
].reset_index(drop=True)

print("RAW batches needing DOE rows:", len(raw_missing))
print("DOE rows needing corrected IDs:", len(doe_wrong))

print("\nRAW IDs:")
print(raw_missing_ids)

print("\nWrong DOE IDs:")
print(doe_wrong_ids)
```

```python
from scipy.optimize import linear_sum_assignment

def robust_pair_cost(raw_row, doe_row):

    diffs = []

    comparisons = [
        # RAW column, DOE column, tolerance/scale

        ("charge_temp", "Actual Charge Temp (°C)", 0.5),

        ("tp_temp", "Actual TP Temp (°C)", 1.0),
        ("yellow_temp", "Actual Yellow Temp (°C)", 0.5),
        ("fc_temp", "Actual FC Temp (°C)", 0.5),
        ("drop_temp", "Actual Drop Temp (°C)", 0.25),

        ("tp_time", "tp_time", 5.0),
        ("yellow_time", "yellow_time", 5.0),
        ("fc_time", "fc_time", 5.0),
        ("drop_time", "drop_time", 5.0),
    ]

    for raw_col, doe_col, scale in comparisons:

        r = raw_row[raw_col]
        d = doe_row[doe_col]

        if pd.notna(r) and pd.notna(d):

            # Cap extreme differences so one typo cannot dominate
            normalized = min(abs(r - d) / scale, 10)

            diffs.append(normalized)

    if not diffs:
        return 9999

    # Median is resistant to individual bad entries
    return np.median(diffs)


# Build 25 x 25 matching matrix
cost25 = np.zeros((len(raw_missing), len(doe_wrong)))

for i, raw_row in raw_missing.iterrows():
    for j, doe_row in doe_wrong.iterrows():
        cost25[i, j] = robust_pair_cost(raw_row, doe_row)


# One-to-one assignment
raw_idx, doe_idx = linear_sum_assignment(cost25)

suggested_matches = []

for ri, di in zip(raw_idx, doe_idx):

    r = raw_missing.iloc[ri]
    d = doe_wrong.iloc[di]

    suggested_matches.append({
        "Run #": d["Run #"],
        "DOE_old_ID": int(d["batch_id"]),
        "RAW_true_ID": int(r["batch_id"]),

        "score": round(cost25[ri, di], 3),

        "FC_time_diff": (
            abs(r["fc_time"] - d["fc_time"])
            if pd.notna(r["fc_time"]) and pd.notna(d["fc_time"])
            else np.nan
        ),

        "Drop_time_diff": (
            abs(r["drop_time"] - d["drop_time"])
            if pd.notna(r["drop_time"]) and pd.notna(d["drop_time"])
            else np.nan
        ),

        "FC_temp_diff": (
            abs(r["fc_temp"] - d["Actual FC Temp (°C)"])
            if pd.notna(r["fc_temp"]) and pd.notna(d["Actual FC Temp (°C)"])
            else np.nan
        ),

        "Drop_temp_diff": (
            abs(r["drop_temp"] - d["Actual Drop Temp (°C)"])
            if pd.notna(r["drop_temp"]) and pd.notna(d["Actual Drop Temp (°C)"])
            else np.nan
        )
    })


suggested_matches = (
    pd.DataFrame(suggested_matches)
    .sort_values("Run #")
)

display(suggested_matches)
```

```python
def sec_to_mmss(seconds):
    if pd.isna(seconds):
        return np.nan

    minutes = int(seconds // 60)
    secs = seconds % 60

    return f"{minutes}:{secs:04.1f}"


raw_lookup = raw_reference.set_index("batch_id")

for idx, row in doe_fixed.iterrows():

    batch = int(row["batch_id"])
    r = raw_lookup.loc[batch]

    # Temperatures
    doe_fixed.at[idx, "Actual Charge Temp (°C)"] = r["charge_temp"]
    doe_fixed.at[idx, "Actual TP Temp (°C)"] = r["tp_temp"]
    doe_fixed.at[idx, "Actual Yellow Temp (°C)"] = r["yellow_temp"]
    doe_fixed.at[idx, "Actual FC Temp (°C)"] = r["fc_temp"]
    doe_fixed.at[idx, "Actual Drop Temp (°C)"] = r["drop_temp"]

    # Event times
    doe_fixed.at[idx, "Actual TP Time"] = sec_to_mmss(r["tp_time"])
    doe_fixed.at[idx, "Actual Yellow Time"] = sec_to_mmss(r["yellow_time"])
    doe_fixed.at[idx, "Actual FC Start Time"] = sec_to_mmss(r["fc_time"])
    doe_fixed.at[idx, "Actual Drop Time"] = sec_to_mmss(r["drop_time"])

    # Development time, only if this column exists
    if "Actual Dev Time" in doe_fixed.columns:
        doe_fixed.at[idx, "Actual Dev Time"] = sec_to_mmss(r["dev_time"])


print("Actual values rebuilt from RAW roast files.")
print("Rows updated:", len(doe_fixed))

display(
    doe_fixed[[
        "Run #",
        "ROEST Roast ID",
        "Actual Charge Temp (°C)",
        "Actual TP Time",
        "Actual TP Temp (°C)",
        "Actual Yellow Time",
        "Actual Yellow Temp (°C)",
        "Actual FC Start Time",
        "Actual FC Temp (°C)",
        "Actual Drop Time",
        "Actual Drop Temp (°C)"
    ]].head(10)
)
```

```python
def build_sequence(csv_path):
    df = load_roast_csv(csv_path).copy()

    # milliseconds → seconds
    df["Time_sec"] = df["Time"] / 1000.0

    # Current BT
    bt = df["Bean temp (°C)"].astype(float).values
    time = df["Time_sec"].astype(float).values

    # X uses time + previous BT
    X_seq = np.column_stack([
        time[1:],
        bt[:-1]
    ])

    # y is the next/current BT
    y_seq = bt[1:]

    return X_seq, y_seq


# Test using one TRAINING roast
sample_row = train_df.iloc[0]

X_sample, y_sample = build_sequence(sample_row["csv_file"])

print("Batch:", sample_row["batch_id"])
print("LSTM input shape:", X_sample.shape)
print("Target shape:", y_sample.shape)

print("\nFirst 5 LSTM inputs:")
print(X_sample[:5])

print("\nFirst 5 target BT values:")
print(y_sample[:5])
```

```python
import torch
from torch.nn.utils.rnn import pad_sequence
from torch.utils.data import Dataset, DataLoader


class RoastDataset(Dataset):
    def __init__(self, samples):
        self.samples = samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]

        return {
            "X": torch.tensor(s["X_seq_scaled"], dtype=torch.float32),
            "y": torch.tensor(s["y_seq_scaled"], dtype=torch.float32),
            "static": torch.tensor(s["static_scaled"], dtype=torch.float32),
            "batch_id": s["batch_id"]
        }


def collate_roasts(batch):

    X_list = [item["X"] for item in batch]
    y_list = [item["y"] for item in batch]

    lengths = torch.tensor(
        [len(x) for x in X_list],
        dtype=torch.long
    )

    # Pad sequences to longest roast in current batch
    X_padded = pad_sequence(
        X_list,
        batch_first=True,
        padding_value=0.0
    )

    y_padded = pad_sequence(
        y_list,
        batch_first=True,
        padding_value=0.0
    )

    static = torch.stack([
        item["static"] for item in batch
    ])

    # True = real timestep
    # False = padding
    max_len = X_padded.shape[1]

    mask = (
        torch.arange(max_len)[None, :]
        < lengths[:, None]
    )

    batch_ids = [
        item["batch_id"] for item in batch
    ]

    return X_padded, static, y_padded, mask, lengths, batch_ids


# Separate datasets
train_samples = [s for s in all_samples if s["split"] == "train"]
val_samples = [s for s in all_samples if s["split"] == "validation"]
test_samples = [s for s in all_samples if s["split"] == "test"]

train_dataset = RoastDataset(train_samples)
val_dataset = RoastDataset(val_samples)
test_dataset = RoastDataset(test_samples)

# Temporary loader for testing
train_loader = DataLoader(
    train_dataset,
    batch_size=8,
    shuffle=True,
    collate_fn=collate_roasts
)


# Test one batch
X_batch, static_batch, y_batch, mask_batch, lengths, batch_ids = next(iter(train_loader))

print("X batch:", X_batch.shape)
print("Static batch:", static_batch.shape)
print("y batch:", y_batch.shape)
print("Mask batch:", mask_batch.shape)

print("\nSequence lengths:")
print(lengths)

print("\nBatch IDs:")
print(batch_ids)

print("\nReal timesteps:", mask_batch.sum().item())
print("Total padded matrix positions:", mask_batch.numel())
```

```python
import torch.nn as nn


class HybridLSTMMLP(nn.Module):
    def __init__(
        self,
        lstm_input_size=2,
        lstm_hidden_size=64,
        static_input_size=4,
        mlp_hidden_size=32,
        dropout=0.2
    ):
        super().__init__()

        # LSTM branch:
        # [elapsed time, previous BT]
        self.lstm = nn.LSTM(
            input_size=lstm_input_size,
            hidden_size=lstm_hidden_size,
            batch_first=True
        )

        # MLP branch:
        # [airflow, planned dev time, preheat, charge temp]
        self.mlp = nn.Sequential(
            nn.Linear(static_input_size, mlp_hidden_size),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(mlp_hidden_size, mlp_hidden_size),
            nn.ReLU()
        )

        # Combine LSTM + MLP information
        self.fusion = nn.Sequential(
            nn.Linear(
                lstm_hidden_size + mlp_hidden_size,
                64
            ),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def forward(self, X_seq, static):

        # LSTM output for every timestep
        lstm_out, _ = self.lstm(X_seq)

        # MLP output once per roast
        static_out = self.mlp(static)

        # Repeat static information across all timesteps
        static_repeated = static_out.unsqueeze(1).expand(
            -1,
            X_seq.size(1),
            -1
        )

        # Join temporal + static features
        combined = torch.cat(
            [lstm_out, static_repeated],
            dim=2
        )

        # Predict BT at every timestep
        prediction = self.fusion(combined)

        return prediction.squeeze(-1)


# Use GPU
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = HybridLSTMMLP().to(device)

print(model)
print("\nDevice:", device)

# Test one batch
X_test = X_batch.to(device)
static_test = static_batch.to(device)

with torch.no_grad():
    pred = model(X_test, static_test)

print("\nInput shape:", X_test.shape)
print("Prediction shape:", pred.shape)
print("Target shape:", y_batch.shape)
```

```python
import torch.optim as optim


def masked_mse_loss(pred, target, mask):
    squared_error = (pred - target) ** 2
    masked_error = squared_error * mask.float()

    return masked_error.sum() / mask.sum()


optimizer = optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)

# Quick test on the current batch
pred = model(
    X_batch.to(device),
    static_batch.to(device)
)

loss = masked_mse_loss(
    pred,
    y_batch.to(device),
    mask_batch.to(device)
)

print("Masked MSE loss:", loss.item())
print("Optimizer:", optimizer)
```

```python
from copy import deepcopy

# Validation loader
val_loader = DataLoader(
    val_dataset,
    batch_size=8,
    shuffle=False,
    collate_fn=collate_roasts
)

max_epochs = 50
patience = 8

best_val_loss = float("inf")
best_model_state = None
epochs_without_improvement = 0

train_history = []
val_history = []

for epoch in range(1, max_epochs + 1):

    # =====================
    # TRAINING
    # =====================
    model.train()

    train_loss_sum = 0.0
    train_points = 0

    for X, static, y, mask, lengths, batch_ids in train_loader:

        X = X.to(device)
        static = static.to(device)
        y = y.to(device)
        mask = mask.to(device)

        optimizer.zero_grad()

        pred = model(X, static)

        loss = masked_mse_loss(
            pred,
            y,
            mask
        )

        loss.backward()

        # Helps prevent unstable LSTM gradients
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        real_points = mask.sum().item()

        train_loss_sum += loss.item() * real_points
        train_points += real_points

    train_loss = train_loss_sum / train_points


    # =====================
    # VALIDATION
    # =====================
    model.eval()

    val_loss_sum = 0.0
    val_points = 0

    with torch.no_grad():

        for X, static, y, mask, lengths, batch_ids in val_loader:

            X = X.to(device)
            static = static.to(device)
            y = y.to(device)
            mask = mask.to(device)

            pred = model(X, static)

            loss = masked_mse_loss(
                pred,
                y,
                mask
            )

            real_points = mask.sum().item()

            val_loss_sum += loss.item() * real_points
            val_points += real_points

    val_loss = val_loss_sum / val_points

    train_history.append(train_loss)
    val_history.append(val_loss)

    print(
        f"Epoch {epoch:02d} | "
        f"Train MSE: {train_loss:.6f} | "
        f"Val MSE: {val_loss:.6f}"
    )


    # =====================
    # EARLY STOPPING
    # =====================
    if val_loss < best_val_loss:

        best_val_loss = val_loss
        best_model_state = deepcopy(model.state_dict())

        epochs_without_improvement = 0

    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= patience:
        print("\nEarly stopping triggered.")
        break


# Restore best validation model
model.load_state_dict(best_model_state)

print("\nTraining complete.")
print("Best validation MSE:", best_val_loss)
```

```python
def recursive_predict(sample, model, device):
    model.eval()

    X_original = sample["X_seq"]
    static_scaled = torch.tensor(
        sample["static_scaled"],
        dtype=torch.float32,
        device=device
    ).unsqueeze(0)

    # Static MLP features calculated once
    with torch.no_grad():
        static_out = model.mlp(static_scaled)

        hidden = None

        # First known BT = initial BT of roast
        prev_bt = X_original[0, 1]

        predictions = []

        for t in range(len(X_original)):

            # Current elapsed time
            time_sec = X_original[t, 0]

            # Scale current time
            time_scaled = time_scaler.transform(
                [[time_sec]]
            )[0, 0]

            # Scale previous predicted BT
            prev_bt_scaled = bt_scaler.transform(
                [[prev_bt]]
            )[0, 0]

            x_t = torch.tensor(
                [[[time_scaled, prev_bt_scaled]]],
                dtype=torch.float32,
                device=device
            )

            # LSTM one timestep at a time
            lstm_out, hidden = model.lstm(
                x_t,
                hidden
            )

            combined = torch.cat(
                [lstm_out[:, 0, :], static_out],
                dim=1
            )

            pred_scaled = model.fusion(
                combined
            ).item()

            # Convert prediction back to Celsius
            pred_bt = bt_scaler.inverse_transform(
                [[pred_scaled]]
            )[0, 0]

            predictions.append(pred_bt)

            # IMPORTANT:
            # next timestep receives MODEL prediction
            prev_bt = pred_bt

    return np.array(predictions)


recursive_predictions = []
recursive_true = []

for sample in test_samples:

    pred = recursive_predict(
        sample,
        model,
        device
    )

    true = sample["y_seq"]

    recursive_predictions.extend(pred)
    recursive_true.extend(true)


recursive_predictions = np.array(recursive_predictions)
recursive_true = np.array(recursive_true)

recursive_mae = mean_absolute_error(
    recursive_true,
    recursive_predictions
)

recursive_mse = mean_squared_error(
    recursive_true,
    recursive_predictions
)

recursive_rmse = np.sqrt(recursive_mse)

print("RECURSIVE TEST RESULTS")
print("-------------------------")
print(f"MAE:  {recursive_mae:.4f} °C")
print(f"MSE:  {recursive_mse:.4f} °C²")
print(f"RMSE: {recursive_rmse:.4f} °C")
print(f"Temperature points: {len(recursive_true)}")
```

```python
def recursive_rollout_batch(
    model,
    X,
    static,
    y,
    teacher_forcing_ratio=0.0
):
    batch_size = X.size(0)
    seq_len = X.size(1)

    # Static MLP branch once per roast
    static_out = model.mlp(static)

    hidden = None

    # Initial BT is known
    prev_bt = X[:, 0, 1]

    predictions = []

    for t in range(seq_len):

        time_t = X[:, t, 0]

        x_t = torch.stack(
            [time_t, prev_bt],
            dim=1
        ).unsqueeze(1)

        lstm_out, hidden = model.lstm(
            x_t,
            hidden
        )

        combined = torch.cat(
            [
                lstm_out[:, 0, :],
                static_out
            ],
            dim=1
        )

        pred_t = model.fusion(
            combined
        ).squeeze(-1)

        predictions.append(pred_t)

        # Input BT for next timestep
        if t < seq_len - 1:

            if teacher_forcing_ratio > 0:

                use_real = (
                    torch.rand(
                        batch_size,
                        device=X.device
                    )
                    < teacher_forcing_ratio
                )

                # y[:, t] becomes previous BT at t+1
                prev_bt = torch.where(
                    use_real,
                    y[:, t],
                    pred_t.detach()
                )

            else:
                prev_bt = pred_t.detach()

    return torch.stack(
        predictions,
        dim=1
    )
```

```python
# Reload baseline weights first
model.load_state_dict(
    torch.load(
        MODEL_DIR / "hybrid_lstm_mlp_baseline.pt",
        map_location=device
    )
)

recursive_optimizer = optim.Adam(
    model.parameters(),
    lr=0.0003,
    weight_decay=1e-5
)

max_recursive_epochs = 25
patience = 6

best_recursive_val = float("inf")
best_recursive_state = None
no_improvement = 0

for epoch in range(1, max_recursive_epochs + 1):

    # Gradually reduce teacher forcing
    teacher_ratio = max(
        0.10,
        0.80 - ((epoch - 1) / (max_recursive_epochs - 1)) * 0.70
    )

    # ====================
    # TRAIN
    # ====================
    model.train()

    train_sum = 0
    train_points = 0

    for X, static, y, mask, lengths, batch_ids in train_loader:

        X = X.to(device)
        static = static.to(device)
        y = y.to(device)
        mask = mask.to(device)

        recursive_optimizer.zero_grad()

        pred = recursive_rollout_batch(
            model,
            X,
            static,
            y,
            teacher_forcing_ratio=teacher_ratio
        )

        loss = masked_mse_loss(
            pred,
            y,
            mask
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        recursive_optimizer.step()

        points = mask.sum().item()

        train_sum += loss.item() * points
        train_points += points

    train_loss = train_sum / train_points


    # ====================
    # VALIDATION
    # fully recursive
    # ====================
    model.eval()

    val_sum = 0
    val_points = 0

    with torch.no_grad():

        for X, static, y, mask, lengths, batch_ids in val_loader:

            X = X.to(device)
            static = static.to(device)
            y = y.to(device)
            mask = mask.to(device)

            pred = recursive_rollout_batch(
                model,
                X,
                static,
                y,
                teacher_forcing_ratio=0.0
            )

            loss = masked_mse_loss(
                pred,
                y,
                mask
            )

            points = mask.sum().item()

            val_sum += loss.item() * points
            val_points += points

    val_loss = val_sum / val_points

    print(
        f"Epoch {epoch:02d} | "
        f"Teacher: {teacher_ratio:.2f} | "
        f"Train: {train_loss:.6f} | "
        f"Recursive Val: {val_loss:.6f}"
    )

    if val_loss < best_recursive_val:

        best_recursive_val = val_loss
        best_recursive_state = deepcopy(
            model.state_dict()
        )

        no_improvement = 0

    else:
        no_improvement += 1

    if no_improvement >= patience:
        print("\nEarly stopping triggered.")
        break


model.load_state_dict(best_recursive_state)

print("\nRecursive fine-tuning complete.")
print("Best recursive validation MSE:", best_recursive_val)
```

```python
import pyswarms as ps

# Current roast settings x_k
current_airflow = 71.0
current_dev = 59.0

lambda_penalty = 0.01


def pso_objective(particles):

    costs = []

    for particle in particles:

        airflow = particle[0]
        dev_time = particle[1]

        # Predicted target-profile MSE
        mse, _ = target_error_for_settings(
            sample,
            airflow,
            dev_time
        )

        # Adaptive penalty:
        # lambda * ||x - x_k||^2
        penalty = lambda_penalty * (
            (airflow - current_airflow) ** 2 +
            (dev_time - current_dev) ** 2
        )

        fitness = mse + penalty

        costs.append(fitness)

    return np.array(costs)


# PSO bounds
lower_bounds = np.array([50.0, 30.0])
upper_bounds = np.array([100.0, 90.0])

bounds = (
    lower_bounds,
    upper_bounds
)

options = {
    "c1": 1.5,
    "c2": 1.5,
    "w": 0.7
}

optimizer_pso = ps.single.GlobalBestPSO(
    n_particles=20,
    dimensions=2,
    options=options,
    bounds=bounds
)

best_cost, best_position = optimizer_pso.optimize(
    pso_objective,
    iters=30,
    verbose=True
)

best_airflow = best_position[0]
best_dev = best_position[1]

best_mse, best_rmse = target_error_for_settings(
    sample,
    best_airflow,
    best_dev
)

print("\nPSO RESULT")
print("-----------------------------")
print(f"Current Airflow: {current_airflow:.1f}%")
print(f"Current Dev Time: {current_dev:.1f}s")

print(f"\nRecommended Airflow: {best_airflow:.2f}%")
print(f"Recommended Dev Time: {best_dev:.2f}s")

print(f"\nCurrent Target MSE: 127.6453 °C²")
print(f"Recommended Target MSE: {best_mse:.4f} °C²")
print(f"Recommended Target RMSE: {best_rmse:.4f} °C")

print(f"\nPSO fitness including penalty: {best_cost:.4f}")
```

```python
def predict_with_settings_v2(sample, airflow, dev_time, first_crack_time):

    # Candidate drop time
    candidate_drop = int(round(first_crack_time + dev_time))

    # Generate elapsed seconds from 1s until candidate drop
    candidate_times = np.arange(
        1,
        candidate_drop + 1,
        dtype=np.float32
    )

    # Static features
    new_static = np.array([
        airflow,
        dev_time,
        sample["static"][2],   # preheat
        sample["static"][3]    # charge temp
    ], dtype=np.float32)

    static_scaled = static_scaler.transform(
        new_static.reshape(1, -1)
    ).flatten().astype(np.float32)

    static_tensor = torch.tensor(
        static_scaled,
        dtype=torch.float32,
        device=device
    ).unsqueeze(0)

    model.eval()

    predictions = []

    # Known BT at t=0
    prev_bt = float(sample["X_seq"][0, 1])

    hidden = None

    with torch.no_grad():

        static_out = model.mlp(static_tensor)

        for time_sec in candidate_times:

            time_scaled = time_scaler.transform(
                [[time_sec]]
            )[0, 0]

            prev_bt_scaled = bt_scaler.transform(
                [[prev_bt]]
            )[0, 0]

            x_t = torch.tensor(
                [[[time_scaled, prev_bt_scaled]]],
                dtype=torch.float32,
                device=device
            )

            lstm_out, hidden = model.lstm(
                x_t,
                hidden
            )

            combined = torch.cat(
                [lstm_out[:, 0, :], static_out],
                dim=1
            )

            pred_scaled = model.fusion(
                combined
            ).item()

            pred_bt = bt_scaler.inverse_transform(
                [[pred_scaled]]
            )[0, 0]

            predictions.append(pred_bt)

            prev_bt = pred_bt

    return candidate_times, np.array(predictions)
```

```python
TARGET_END_TIME = 480


def target_error_v2(sample, airflow, dev_time, first_crack_time):

    times, pred = predict_with_settings_v2(
        sample,
        airflow,
        dev_time,
        first_crack_time
    )

    # Only compare where the official target exists
    valid = times <= TARGET_END_TIME

    compare_times = times[valid]
    compare_pred = pred[valid]

    target = target_interpolator(compare_times)

    mse = mean_squared_error(
        target,
        compare_pred
    )

    rmse = np.sqrt(mse)

    candidate_drop = times[-1]

    # Difference from target roast duration
    duration_error = abs(
        candidate_drop - TARGET_END_TIME
    )

    return mse, rmse, candidate_drop, duration_error
```

```python
lambda_change = 0.01
beta_duration = 0.005

current_airflow = 71.0
current_dev = 59.0
first_crack_time = 427.0


def pso_objective_v2(particles):

    costs = []

    for airflow, dev_time in particles:

        mse, rmse, drop, duration_error = target_error_v2(
            sample,
            airflow,
            dev_time,
            first_crack_time
        )

        # Penalize large changes from current roast settings
        change_penalty = lambda_change * (
            (airflow - current_airflow) ** 2 +
            (dev_time - current_dev) ** 2
        )

        # Penalize candidate drop time differing from
        # the target profile endpoint (480 s)
        duration_penalty = beta_duration * (
            duration_error ** 2
        )

        fitness = (
            mse
            + change_penalty
            + duration_penalty
        )

        costs.append(fitness)

    return np.array(costs)


# Check fitness of the four settings we just tested
for airflow, dev in [
    (71, 59),
    (50, 30),
    (100, 90),
    (64.74, 68.26)
]:

    fitness = pso_objective_v2(
        np.array([[airflow, dev]])
    )[0]

    print(
        f"Airflow={airflow:6.2f}% | "
        f"Dev={dev:5.2f}s | "
        f"Fitness={fitness:.3f}"
    )
```

```python
def pso_objective_fast(particles):

    n_particles = len(particles)

    airflows = particles[:, 0]
    dev_times = particles[:, 1]

    # Candidate drop times
    drop_times = np.rint(
        first_crack_time + dev_times
    ).astype(int)

    max_drop = drop_times.max()

    # --------------------------------
    # Static features for all particles
    # --------------------------------
    static_values = np.column_stack([
        airflows,
        dev_times,
        np.full(n_particles, sample["static"][2]),
        np.full(n_particles, sample["static"][3])
    ]).astype(np.float32)

    static_scaled = (
        (static_values - static_scaler.mean_)
        / static_scaler.scale_
    ).astype(np.float32)

    static_tensor = torch.tensor(
        static_scaled,
        dtype=torch.float32,
        device=device
    )

    # Initial BT at t = 0
    initial_bt = float(sample["X_seq"][0, 1])

    prev_bt = torch.full(
        (n_particles,),
        initial_bt,
        dtype=torch.float32,
        device=device
    )

    hidden = None
    predictions = []

    model.eval()

    with torch.no_grad():

        static_out = model.mlp(static_tensor)

        for t in range(1, max_drop + 1):

            # Scale time directly
            time_scaled = (
                (t - time_scaler.mean_[0])
                / time_scaler.scale_[0]
            )

            time_tensor = torch.full(
                (n_particles,),
                float(time_scaled),
                dtype=torch.float32,
                device=device
            )

            # Scale previous BT directly
            prev_bt_scaled = (
                (prev_bt - bt_scaler.mean_[0])
                / bt_scaler.scale_[0]
            )

            x_t = torch.stack(
                [time_tensor, prev_bt_scaled],
                dim=1
            ).unsqueeze(1)

            lstm_out, hidden = model.lstm(
                x_t,
                hidden
            )

            combined = torch.cat(
                [lstm_out[:, 0, :], static_out],
                dim=1
            )

            pred_scaled = model.fusion(
                combined
            ).squeeze(-1)

            # Convert back to °C
            pred_bt = (
                pred_scaled * bt_scaler.scale_[0]
                + bt_scaler.mean_[0]
            )

            predictions.append(pred_bt.cpu().numpy())

            prev_bt = pred_bt

    # shape = [time, particle]
    predictions = np.stack(predictions)

    costs = []

    for i in range(n_particles):

        drop = drop_times[i]

        compare_end = min(
            drop,
            TARGET_END_TIME
        )

        pred_valid = predictions[
            :compare_end,
            i
        ]

        compare_times = np.arange(
            1,
            compare_end + 1
        )

        target_valid = target_interpolator(
            compare_times
        )

        mse = np.mean(
            (pred_valid - target_valid) ** 2
        )

        duration_error = abs(
            drop - TARGET_END_TIME
        )

        change_penalty = lambda_change * (
            (airflows[i] - current_airflow) ** 2
            +
            (dev_times[i] - current_dev) ** 2
        )

        duration_penalty = beta_duration * (
            duration_error ** 2
        )

        costs.append(
            mse
            + change_penalty
            + duration_penalty
        )

    return np.array(costs)
```

```python
def make_fast_pso_objective(
    roast_sample,
    current_airflow_value,
    current_dev_value,
    fc_time_value
):

    def objective(particles):

        n_particles = len(particles)

        airflows = particles[:, 0]
        dev_times = particles[:, 1]

        drop_times = np.rint(
            fc_time_value + dev_times
        ).astype(int)

        max_drop = drop_times.max()

        # Static features
        static_values = np.column_stack([
            airflows,
            dev_times,
            np.full(n_particles, roast_sample["static"][2]),
            np.full(n_particles, roast_sample["static"][3])
        ]).astype(np.float32)

        static_scaled = (
            (static_values - static_scaler.mean_)
            / static_scaler.scale_
        ).astype(np.float32)

        static_tensor = torch.tensor(
            static_scaled,
            dtype=torch.float32,
            device=device
        )

        initial_bt = float(
            roast_sample["X_seq"][0, 1]
        )

        prev_bt = torch.full(
            (n_particles,),
            initial_bt,
            dtype=torch.float32,
            device=device
        )

        hidden = None
        predictions = []

        model.eval()

        with torch.no_grad():

            static_out = model.mlp(static_tensor)

            for t in range(1, max_drop + 1):

                time_scaled = (
                    (t - time_scaler.mean_[0])
                    / time_scaler.scale_[0]
                )

                time_tensor = torch.full(
                    (n_particles,),
                    float(time_scaled),
                    dtype=torch.float32,
                    device=device
                )

                prev_bt_scaled = (
                    (prev_bt - bt_scaler.mean_[0])
                    / bt_scaler.scale_[0]
                )

                x_t = torch.stack(
                    [time_tensor, prev_bt_scaled],
                    dim=1
                ).unsqueeze(1)

                lstm_out, hidden = model.lstm(
                    x_t,
                    hidden
                )

                combined = torch.cat(
                    [lstm_out[:, 0, :], static_out],
                    dim=1
                )

                pred_scaled = model.fusion(
                    combined
                ).squeeze(-1)

                pred_bt = (
                    pred_scaled * bt_scaler.scale_[0]
                    + bt_scaler.mean_[0]
                )

                predictions.append(
                    pred_bt.cpu().numpy()
                )

                prev_bt = pred_bt

        predictions = np.stack(predictions)

        costs = []

        for i in range(n_particles):

            drop = drop_times[i]

            compare_end = min(
                drop,
                TARGET_END_TIME
            )

            pred_valid = predictions[:compare_end, i]

            compare_times = np.arange(
                1,
                compare_end + 1
            )

            target_valid = target_interpolator(
                compare_times
            )

            mse = np.mean(
                (pred_valid - target_valid) ** 2
            )

            duration_error = abs(
                drop - TARGET_END_TIME
            )

            change_penalty = lambda_change * (
                (airflows[i] - current_airflow_value) ** 2
                +
                (dev_times[i] - current_dev_value) ** 2
            )

            duration_penalty = beta_duration * (
                duration_error ** 2
            )

            costs.append(
                mse
                + change_penalty
                + duration_penalty
            )

        return np.array(costs)

    return objective


print("Reusable PSO objective created.")
```

```python
all_pso_results = []

for sample_i in test_samples:

    batch_id = sample_i["batch_id"]

    # Find matching CSV
    csv_path = corrected_doe.loc[
        corrected_doe["batch_id"] == batch_id,
        "csv_file"
    ].iloc[0]

    df = load_roast_csv(csv_path).copy()
    df["Time_sec"] = df["Time"] / 1000

    # First Crack time
    fc_time = df.loc[
        df["Event"] == "First Crack",
        "Time_sec"
    ].iloc[0]

    current_airflow_i = float(sample_i["static"][0])
    current_dev_i = float(sample_i["static"][1])

    # Current error
    current_mse, current_rmse, current_drop, _ = target_error_v2(
        sample_i,
        current_airflow_i,
        current_dev_i,
        fc_time
    )

    # Create PSO objective for this roast
    objective_i = make_fast_pso_objective(
        sample_i,
        current_airflow_i,
        current_dev_i,
        fc_time
    )

    optimizer_i = ps.single.GlobalBestPSO(
        n_particles=20,
        dimensions=2,
        options={
            "c1": 1.5,
            "c2": 1.5,
            "w": 0.7
        },
        bounds=(
            np.array([50.0, 30.0]),
            np.array([100.0, 90.0])
        )
    )

    start = time.time()

    best_cost, best_pos = optimizer_i.optimize(
        objective_i,
        iters=30,
        verbose=False
    )

    runtime = time.time() - start

    # Recommended error
    rec_mse, rec_rmse, rec_drop, _ = target_error_v2(
        sample_i,
        best_pos[0],
        best_pos[1],
        fc_time
    )

    improvement = (
        (current_mse - rec_mse)
        / current_mse
    ) * 100

    all_pso_results.append({
        "batch_id": batch_id,
        "current_airflow": current_airflow_i,
        "recommended_airflow": best_pos[0],
        "current_dev": current_dev_i,
        "recommended_dev": best_pos[1],
        "current_rmse": current_rmse,
        "recommended_rmse": rec_rmse,
        "current_mse": current_mse,
        "recommended_mse": rec_mse,
        "improvement_pct": improvement,
        "runtime_sec": runtime
    })

    print(
        f"Batch {batch_id} done | "
        f"Improvement: {improvement:.2f}%"
    )


all_pso_results = pd.DataFrame(all_pso_results)

display(all_pso_results)

print("\nSUMMARY")
print("-------------------------")
print(
    "Improved roasts:",
    (all_pso_results["improvement_pct"] > 0).sum(),
    "/ 12"
)

print(
    "Average improvement:",
    all_pso_results["improvement_pct"].mean()
)

print(
    "Median improvement:",
    all_pso_results["improvement_pct"].median()
)
```

```python
class StandardLSTM(nn.Module):
    def __init__(
        self,
        input_size=2,
        hidden_size=64,
        dropout=0.2
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.output = nn.Sequential(
            nn.Linear(hidden_size, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def forward(self, X_seq):

        lstm_out, _ = self.lstm(X_seq)

        prediction = self.output(
            lstm_out
        )

        return prediction.squeeze(-1)


baseline_lstm = StandardLSTM().to(device)

print(baseline_lstm)
print("\nDevice:", device)

# Quick shape test
with torch.no_grad():
    baseline_pred = baseline_lstm(
        X_batch.to(device)
    )

print("\nPrediction shape:", baseline_pred.shape)
print("Target shape:", y_batch.shape)
```

```python
def recursive_rollout_standard(model, X, teacher_ratio=0.0):

    batch_size, seq_len, _ = X.shape

    hidden = None
    preds = []

    # Real initial BT
    prev_bt = X[:, 0, 1]

    for t in range(seq_len):

        time_t = X[:, t, 0]

        # From t > 0, decide whether to use
        # true previous BT or model prediction
        if t > 0 and teacher_ratio > 0:

            true_prev_bt = X[:, t, 1]

            use_teacher = (
                torch.rand(batch_size, device=X.device)
                < teacher_ratio
            )

            prev_bt = torch.where(
                use_teacher,
                true_prev_bt,
                prev_pred
            )

        input_t = torch.stack(
            [time_t, prev_bt],
            dim=1
        ).unsqueeze(1)

        lstm_out, hidden = model.lstm(
            input_t,
            hidden
        )

        pred_t = model.output(
            lstm_out
        ).squeeze(1).squeeze(-1)

        preds.append(pred_t)

        prev_pred = pred_t
        prev_bt = pred_t

    return torch.stack(preds, dim=1)


# Fine-tuning optimizer
baseline_recursive_optimizer = optim.Adam(
    baseline_lstm.parameters(),
    lr=0.0003,
    weight_decay=1e-5
)

max_epochs = 25
patience = 6

best_recursive_val = float("inf")
best_recursive_state = None
no_improvement = 0

for epoch in range(1, max_epochs + 1):

    # Teacher forcing: 0.80 -> 0.10
    teacher_ratio = max(
        0.10,
        0.80 - ((epoch - 1) / (max_epochs - 1)) * 0.70
    )

    # =========================
    # TRAIN
    # =========================
    baseline_lstm.train()

    train_sum = 0.0
    train_points = 0

    for X, static, y, mask, lengths, batch_ids in train_loader:

        X = X.to(device)
        y = y.to(device)
        mask = mask.to(device)

        baseline_recursive_optimizer.zero_grad()

        pred = recursive_rollout_standard(
            baseline_lstm,
            X,
            teacher_ratio=teacher_ratio
        )

        loss = masked_mse_loss(
            pred,
            y,
            mask
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            baseline_lstm.parameters(),
            max_norm=1.0
        )

        baseline_recursive_optimizer.step()

        points = mask.sum().item()

        train_sum += loss.item() * points
        train_points += points

    train_loss = train_sum / train_points

    # =========================
    # VALIDATION - FULL RECURSIVE
    # =========================
    baseline_lstm.eval()

    val_sum = 0.0
    val_points = 0

    with torch.no_grad():

        for X, static, y, mask, lengths, batch_ids in val_loader:

            X = X.to(device)
            y = y.to(device)
            mask = mask.to(device)

            pred = recursive_rollout_standard(
                baseline_lstm,
                X,
                teacher_ratio=0.0
            )

            loss = masked_mse_loss(
                pred,
                y,
                mask
            )

            points = mask.sum().item()

            val_sum += loss.item() * points
            val_points += points

    val_loss = val_sum / val_points

    print(
        f"Epoch {epoch:02d} | "
        f"Teacher: {teacher_ratio:.2f} | "
        f"Train: {train_loss:.6f} | "
        f"Val Recursive: {val_loss:.6f}"
    )

    if val_loss < best_recursive_val:

        best_recursive_val = val_loss
        best_recursive_state = deepcopy(
            baseline_lstm.state_dict()
        )

        no_improvement = 0

    else:
        no_improvement += 1

    if no_improvement >= patience:
        print("\nEarly stopping triggered.")
        break


baseline_lstm.load_state_dict(
    best_recursive_state
)

print("\nStandard LSTM recursive fine-tuning complete.")
print("Best recursive validation MSE:", best_recursive_val)
```

```python
from sklearn.model_selection import ParameterSampler
from copy import deepcopy

# ==========================================================
# TUNABLE HYBRID MODEL
# ==========================================================

class TunableHybridLSTMMLP(nn.Module):

    def __init__(
        self,
        lstm_hidden=64,
        mlp_hidden=32,
        fusion_hidden=64,
        dropout=0.2
    ):
        super().__init__()

        self.lstm_hidden = lstm_hidden

        self.lstm = nn.LSTM(
            input_size=2,
            hidden_size=lstm_hidden,
            batch_first=True
        )

        self.mlp = nn.Sequential(
            nn.Linear(4, mlp_hidden),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(mlp_hidden, mlp_hidden),
            nn.ReLU()
        )

        self.fusion = nn.Sequential(
            nn.Linear(
                lstm_hidden + mlp_hidden,
                fusion_hidden
            ),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(fusion_hidden, 1)
        )

    def forward(self, X_seq, static):

        lstm_out, _ = self.lstm(X_seq)

        static_features = self.mlp(static)

        static_features = static_features.unsqueeze(1).expand(
            -1,
            X_seq.size(1),
            -1
        )

        combined = torch.cat(
            [lstm_out, static_features],
            dim=2
        )

        prediction = self.fusion(combined)

        return prediction.squeeze(-1)


# ==========================================================
# RECURSIVE ROLLOUT
# ==========================================================

def recursive_rollout_hybrid_tuned(
    model,
    X,
    static,
    teacher_ratio=0.0
):

    batch_size, seq_len, _ = X.shape

    hidden = None
    preds = []

    static_features = model.mlp(static)

    prev_bt = X[:, 0, 1]

    for t in range(seq_len):

        time_t = X[:, t, 0]

        if t > 0 and teacher_ratio > 0:

            true_prev_bt = X[:, t, 1]

            use_teacher = (
                torch.rand(
                    batch_size,
                    device=X.device
                ) < teacher_ratio
            )

            prev_bt = torch.where(
                use_teacher,
                true_prev_bt,
                prev_pred
            )

        input_t = torch.stack(
            [time_t, prev_bt],
            dim=1
        ).unsqueeze(1)

        lstm_out, hidden = model.lstm(
            input_t,
            hidden
        )

        lstm_features = lstm_out.squeeze(1)

        combined = torch.cat(
            [lstm_features, static_features],
            dim=1
        )

        pred_t = model.fusion(
            combined
        ).squeeze(-1)

        preds.append(pred_t)

        prev_pred = pred_t
        prev_bt = pred_t

    return torch.stack(preds, dim=1)


# ==========================================================
# RANDOM SEARCH SPACE
# ==========================================================

param_grid = {
    "lstm_hidden": [32, 64, 96],
    "mlp_hidden": [16, 32, 64],
    "fusion_hidden": [32, 64],
    "dropout": [0.1, 0.2, 0.3],
    "lr": [0.0003, 0.0005, 0.001],
    "weight_decay": [1e-5, 1e-4]
}

search_configs = list(
    ParameterSampler(
        param_grid,
        n_iter=10,
        random_state=42
    )
)

random_search_results = []

best_overall_val = float("inf")
best_overall_state = None
best_overall_config = None


# ==========================================================
# RANDOM SEARCH
# ==========================================================

for trial, config in enumerate(
    search_configs,
    start=1
):

    print(
        f"\n========== TRIAL {trial}/10 =========="
    )

    print(config)

    # Reproducible initialization
    torch.manual_seed(42 + trial)
    np.random.seed(42 + trial)

    model_trial = TunableHybridLSTMMLP(
        lstm_hidden=config["lstm_hidden"],
        mlp_hidden=config["mlp_hidden"],
        fusion_hidden=config["fusion_hidden"],
        dropout=config["dropout"]
    ).to(device)

    optimizer_trial = optim.Adam(
        model_trial.parameters(),
        lr=config["lr"],
        weight_decay=config["weight_decay"]
    )

    # ======================================================
    # PHASE 1 — NORMAL TRAINING
    # ======================================================

    best_base_val = float("inf")
    best_base_state = None

    patience = 5
    no_improve = 0

    for epoch in range(1, 31):

        model_trial.train()

        for X, static, y, mask, lengths, batch_ids in train_loader:

            X = X.to(device)
            static = static.to(device)
            y = y.to(device)
            mask = mask.to(device)

            optimizer_trial.zero_grad()

            pred = model_trial(
                X,
                static
            )

            loss = masked_mse_loss(
                pred,
                y,
                mask
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model_trial.parameters(),
                1.0
            )

            optimizer_trial.step()

        # VALIDATION
        model_trial.eval()

        val_sum = 0.0
        val_points = 0

        with torch.no_grad():

            for X, static, y, mask, lengths, batch_ids in val_loader:

                X = X.to(device)
                static = static.to(device)
                y = y.to(device)
                mask = mask.to(device)

                pred = model_trial(
                    X,
                    static
                )

                val_loss = masked_mse_loss(
                    pred,
                    y,
                    mask
                )

                points = mask.sum().item()

                val_sum += (
                    val_loss.item()
                    * points
                )

                val_points += points

        val_mse = val_sum / val_points

        if val_mse < best_base_val:

            best_base_val = val_mse

            best_base_state = deepcopy(
                model_trial.state_dict()
            )

            no_improve = 0

        else:
            no_improve += 1

        if no_improve >= patience:
            break


    model_trial.load_state_dict(
        best_base_state
    )


    # ======================================================
    # PHASE 2 — RECURSIVE FINE-TUNING
    # ======================================================

    recursive_optimizer = optim.Adam(
        model_trial.parameters(),
        lr=0.0003,
        weight_decay=config["weight_decay"]
    )

    best_recursive_val = float("inf")
    best_recursive_state = None

    no_improve = 0
    recursive_patience = 4

    for epoch in range(1, 16):

        teacher_ratio = max(
            0.10,
            0.80 -
            ((epoch - 1) / 14) * 0.70
        )

        model_trial.train()

        for X, static, y, mask, lengths, batch_ids in train_loader:

            X = X.to(device)
            static = static.to(device)
            y = y.to(device)
            mask = mask.to(device)

            recursive_optimizer.zero_grad()

            pred = recursive_rollout_hybrid_tuned(
                model_trial,
                X,
                static,
                teacher_ratio
            )

            loss = masked_mse_loss(
                pred,
                y,
                mask
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model_trial.parameters(),
                1.0
            )

            recursive_optimizer.step()


        # FULL RECURSIVE VALIDATION
        model_trial.eval()

        val_sum = 0.0
        val_points = 0

        with torch.no_grad():

            for X, static, y, mask, lengths, batch_ids in val_loader:

                X = X.to(device)
                static = static.to(device)
                y = y.to(device)
                mask = mask.to(device)

                pred = recursive_rollout_hybrid_tuned(
                    model_trial,
                    X,
                    static,
                    teacher_ratio=0.0
                )

                val_loss = masked_mse_loss(
                    pred,
                    y,
                    mask
                )

                points = mask.sum().item()

                val_sum += (
                    val_loss.item()
                    * points
                )

                val_points += points

        recursive_val_mse = (
            val_sum / val_points
        )

        if recursive_val_mse < best_recursive_val:

            best_recursive_val = (
                recursive_val_mse
            )

            best_recursive_state = deepcopy(
                model_trial.state_dict()
            )

            no_improve = 0

        else:
            no_improve += 1

        if no_improve >= recursive_patience:
            break


    # ======================================================
    # SAVE TRIAL RESULT
    # ======================================================

    print(
        f"Best Recursive Val MSE: "
        f"{best_recursive_val:.6f}"
    )

    result = config.copy()

    result["trial"] = trial
    result["base_val_mse"] = best_base_val
    result["recursive_val_mse"] = best_recursive_val

    random_search_results.append(
        result
    )


    # BEST MODEL SO FAR
    if best_recursive_val < best_overall_val:

        best_overall_val = best_recursive_val

        best_overall_config = (
            config.copy()
        )

        best_overall_state = deepcopy(
            best_recursive_state
        )


# ==========================================================
# RESULTS
# ==========================================================

random_search_results = pd.DataFrame(
    random_search_results
)

random_search_results = (
    random_search_results
    .sort_values("recursive_val_mse")
    .reset_index(drop=True)
)

display(random_search_results)

print("\n========== BEST CONFIGURATION ==========")

print(best_overall_config)

print(
    "\nBest Recursive Validation MSE:",
    best_overall_val
)


# ==========================================================
# SAVE SEARCH RESULTS + BEST MODEL
# ==========================================================

SEARCH_PATH = (
    PROJECT / "results" /
    "hybrid_random_search_results.csv"
)

MODEL_PATH = (
    PROJECT / "models" /
    "hybrid_random_search_best.pt"
)

random_search_results.to_csv(
    SEARCH_PATH,
    index=False
)

torch.save(
    {
        "state_dict": best_overall_state,
        "config": best_overall_config,
        "recursive_val_mse": best_overall_val
    },
    MODEL_PATH
)

print("\nSaved search:")
print(SEARCH_PATH)

print("\nSaved best model:")
print(MODEL_PATH)
```
